```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sb
import warnings
import datetime
warnings.filterwarnings('ignore')
%matplotlib inline
```

```python
weather = pd.read_csv("hourly_weather_delhi.csv", skiprows=3)

print(weather.head())
```

```python
weather.rename(columns={
    "time": "Datetime",
    "temperature_2m (°C)": "Temperature",
    "wind_speed_10m (m/s)": "WindSpeed",
    "wind_direction_10m (°)": "WindDirection",
    "rain (mm)": "Rain",
    "relative_humidity_2m (%)": "Humidity"
}, inplace=True)
```

```python
print(weather.columns.tolist())
```

```python
weather.head()
```

```python
weather["Datetime"] = pd.to_datetime(weather["Datetime"])
```

```python
print(weather.head())
```

```python
weather["Date"] = weather["Datetime"].dt.date
```

```python
weather.head()
```

```python
daily_weather = weather.groupby("Date").agg({
    "Temperature": "mean",
    "Humidity": "mean",
    "WindSpeed": "mean",
    "WindDirection": "mean",
    "Rain": "sum"
}).reset_index()
```

```python
daily_weather.head()
```

```python
print("Hourly rows:", len(weather))
print("Daily rows:", len(daily_weather))
```

## Filter AQI data of delhi from AQI.csv dataset

```python
daily_AQI_delhi = pd.read_csv("AQI.csv")

print(daily_AQI_delhi.head())
```

```python
aqi_delhi = daily_AQI_delhi[daily_AQI_delhi["City"] == "Delhi"].copy()
```

```python
print(aqi_delhi.head())
print(aqi_delhi.shape)
```

```python
aqi_delhi.rename(columns={
    "Datetime": "Date"
}, inplace=True)
```

```python
aqi_delhi.head()
```

```python
print(aqi_delhi["Date"].dtype)
print(daily_weather["Date"].dtype)
```

```python
aqi_delhi["Date"] = pd.to_datetime(aqi_delhi["Date"])
daily_weather["Date"] = pd.to_datetime(daily_weather["Date"])
```

```python
print(aqi_delhi["Date"].dtype)
print(daily_weather["Date"].dtype)
```

```python
print(aqi_delhi["Date"].head())
print(daily_weather["Date"].head())
```

```python
merged = pd.merge(
    aqi_delhi,
    daily_weather,
    left_on="Date",
    right_on="Date",
    how="inner"
)
```

```python
print(merged.head())
```

```python
merged["month"] = merged["Date"].dt.month
merged["day_of_week"] = merged["Date"].dt.dayofweek
merged["day_of_year"] = merged["Date"].dt.dayofyear
```

```python
merged.head()
```

```python
merged.to_csv("delhi_aqi_weather_daily.csv", index=False)
```
